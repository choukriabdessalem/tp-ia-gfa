```python

# =====================================================
# TP : REGRESSION LINEAIRE - CROISSANCE BACTERIENNE
# =====================================================

# 1. Importation des bibliothèques
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# 2. Introduction des données avec Pandas
data = {
    'Temperature': [20, 25, 30, 35, 40],
    'Croissance_bacterienne': [40, 55, 70, 85, 100]
}

df = pd.DataFrame(data)

print("1. TABLEAU DES DONNEES")
print(df)

# 3. Visualisation avec un nuage de points
plt.figure(figsize=(8, 5))
plt.scatter(df['Temperature'], df['Croissance_bacterienne'],
            color='blue', label='Donnees experimentales')

plt.xlabel('Temperature (°C)')
plt.ylabel('Croissance bacterienne')
plt.title('Relation entre temperature et croissance bacterienne')
plt.legend()
plt.grid()
plt.show()

# 4. Construction du modele de regression
X = df[['Temperature']]
y = df['Croissance_bacterienne']

model = LinearRegression()
model.fit(X, y)

# 5. Affichage des coefficients
a = model.coef_[0]
b = model.intercept_

print("\n2. COEFFICIENTS DU MODELE")
print("Coefficient directeur (a) :", a)
print("Ordonnee a l'origine (b) :", b)
print("Equation de regression : y =", a, "x +", b)

# 6. Prediction a 48°C
prediction_48 = model.predict(
    pd.DataFrame({'Temperature': [48]})
)

print("\n3. PREDICTION A 48°C")
print("Croissance bacterienne :", prediction_48[0])

# 7. Trace de la droite de regression
y_pred = model.predict(X)

plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', label='Donnees reelles')
plt.plot(X, y_pred, color='red', label='Droite de regression')

plt.xlabel('Temperature (°C)')
plt.ylabel('Croissance bacterienne')
plt.title('Droite de regression lineaire')
plt.legend()
plt.grid()
plt.show()

# 8. Prediction et interpretation a 50°C
prediction_50 = model.predict(
    pd.DataFrame({'Temperature': [50]})
)

print("\n4. PREDICTION A 50°C")
print("Croissance bacterienne predite :", prediction_50[0])
print("Cette prediction est une extrapolation.")

# 9. Evaluation du modele
r2 = r2_score(y, y_pred)
mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)

print("\n5. EVALUATION DU MODELE")
print("Coefficient de determination R² :", r2)
print("Erreur absolue moyenne MAE :", mae)
print("Erreur quadratique moyenne MSE :", mse)
```
